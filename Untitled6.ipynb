```python
from lightgbm import LGBMClassifier
import numpy as np
import pandas as pd
from sklearn.ensemble import (
    AdaBoostClassifier,
    GradientBoostingClassifier,
    RandomForestClassifier,
    VotingClassifier
)
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    f1_score,
    fbeta_score,
    make_scorer,
    precision_score,
    recall_score
)
from sklearn.model_selection import (
    GridSearchCV,
    StratifiedKFold,
    train_test_split
)
from sklearn.neural_network import MLPClassifier
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
import warnings

warnings.filterwarnings("ignore")
```

```python
df= pd.read_csv(filepath_or_buffer="/content/amzn_customer_reviews (1).csv", sep=",")
df.head(3)
```

```python
# 1. Cross-validation setup
cross_validation = StratifiedKFold(n_splits=3, random_state=42, shuffle=True)

# 2. Scoring function using F-beta with beta=3 (prioritizes recall)
fthree_scorer = make_scorer(fbeta_score, average="macro", beta=3)

# 3. Feature and target columns
X = df["Review"]
y = df["Sentiment"]

# 4. Train-test split (75% train, 25% test)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, shuffle=True
)

# Check shapes
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
# Ensure all text is str (not bytes)
X_train = X_train.astype(str)
X_test = X_test.astype(str)

# Vectorization with TF-IDF
vectorizer = TfidfVectorizer(
    encoding="utf-8",
    lowercase=False,
    tokenizer=None,
    analyzer="word",
    stop_words=None,
    ngram_range=(1, 2),
    min_df=1,
    norm="l2",
    use_idf=True
)

# ❌ Avoid this:
# X_train_vec = np.asarray(vectorizer.fit_transform(X_train).todense())

# ✅ Use this:
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)

# Display shape info
print("-" * 65)
print("Train shape: ", X_train_vec.shape)
print("Test shape: ", X_test_vec.shape)
print("-" * 65)

# Check class distribution
print("Originally:")
print(df["Sentiment"].value_counts(normalize=True))
print("\nTrain:")
print(y_train.value_counts(normalize=True))
print("\nTest:")
print(y_test.value_counts(normalize=True))
print("-" * 65)
```

```python
X_train
```

```python
pipeline = Pipeline([
    (
        "log_clf",
        LogisticRegression(
            dual=False,
            tol=0.0001,
            fit_intercept=True,
            intercept_scaling=1,
            class_weight="balanced",
            random_state=42,
            max_iter=100,
            multi_class="auto",
            verbose=False,
            warm_start=False,
            n_jobs=-1
        )
    )
])

# Define hyperparameter grid
param_grid = {
    "log_clf__C": [0.1, 1.0, 10],
    "log_clf__solver": ["liblinear"]  # saga is slower in Colab
}

# GridSearchCV with F3 scorer and cross-validation
from sklearn.model_selection import RandomizedSearchCV

log_grid = RandomizedSearchCV(
    estimator=pipeline,
    param_distributions=param_grid,
    n_iter=3,  # Try 3 combinations only
    cv=cross_validation,
    scoring=fthree_scorer,
    n_jobs=-1,
    random_state=42,
    verbose=True
)


# Fit using the vectorized input
log_grid.fit(X_train_vec, np.ravel(y_train))
```

```python
dict_params = log_grid.best_params_
dict_params
{'log_clf__C': 1.0, 'log_clf__solver': 'saga'}
log_clf = LogisticRegression(
    penalty="l2",
    dual=False,
    tol=0.0001,
    C=dict_params["log_clf__C"],
    fit_intercept=True,
    intercept_scaling=1,
    class_weight="balanced",
    random_state=42,
    solver=dict_params["log_clf__solver"],
    max_iter=100,
    multi_class="auto",
    verbose=False,
    warm_start=False,
    n_jobs=-1,
    l1_ratio=0
)

# Use the vectorized data (not raw text)
log_clf.fit(X_train_vec, np.ravel(y_train))

# Predict using vectorized test set
y_pred = log_clf.predict(X_test_vec)

# Generate confusion matrix
cm = pd.crosstab(y_test, y_pred, rownames=["Real class"], colnames=["Predicted class"])
cm
```

```python
accuracy_log = round(accuracy_score(y_true=y_test, y_pred=y_pred), 4)
recall_log = round(recall_score(y_true=y_test, y_pred=y_pred, average="macro"), 4)
precision_log = round(precision_score(y_true=y_test, y_pred=y_pred, average="macro"), 4)
f1_score_log = round(f1_score(y_true=y_test, y_pred=y_pred, average="macro"), 4)
f3_score_log = round(fbeta_score(y_true=y_test, y_pred=y_pred, average="macro", beta=3), 4)
print(accuracy_log)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.model_selection import GridSearchCV

# Pipeline with RandomForest
pipeline = Pipeline([
    ("rf_clf", RandomForestClassifier(
        n_jobs=-1,
        random_state=42,
        class_weight="balanced"
    ))
])

# Simplified Grid
param_grid = {
    "rf_clf__n_estimators": [100],
    "rf_clf__criterion": ["gini"],
    "rf_clf__max_depth": [10],
    "rf_clf__min_samples_split": [2],
    "rf_clf__min_samples_leaf": [1],
    "rf_clf__max_features": ["sqrt"]
}

# Grid Search
rf_grid = GridSearchCV(
    estimator=pipeline,
    param_grid=param_grid,
    cv=StratifiedKFold(n_splits=2, shuffle=True, random_state=42),  # 2-fold to save time
    scoring=fthree_scorer,
    n_jobs=-1,
    verbose=True
)

# Fit on vectorized sparse data
rf_grid.fit(X_train_vec, np.ravel(y_train))
```

```python
from sklearn.metrics import classification_report, confusion_matrix

best_rf = rf_grid.best_estimator_
y_pred = best_rf.predict(X_test_vec)

print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
```

```python
# Step 3: Define scorer and CV strategy
f3_scorer = make_scorer(fbeta_score, beta=3, average='macro')
cross_validation = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)
```

```python
vectorizer = TfidfVectorizer(ngram_range=(1, 2), lowercase=False)
X = df["Review"]
y = df["Sentiment"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, shuffle=True)

X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)

from sklearn.preprocessing import LabelEncoder

# Encode the target labels
le = LabelEncoder()
y_encoded = le.fit_transform(df["Sentiment"])  # Converts to [0, 1, 2] etc.

# Now split the encoded labels
X = df["Review"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y_encoded, test_size=0.25, random_state=42, stratify=y_encoded
)
```

```python
pipeline = Pipeline([
    ("xgb_clf", xgb.XGBClassifier(
        objective="multi:softmax",  # for direct class prediction
        num_class=len(y.unique()),
        use_label_encoder=False,
        eval_metric="mlogloss",
        tree_method="hist",  # optimized for speed
        n_jobs=-1,
        random_state=42
    ))
])

# Step 6: Define a minimal param grid to keep training fast
param_grid = {
    "xgb_clf__max_depth": [4],
    "xgb_clf__learning_rate": [0.3],
    "xgb_clf__n_estimators": [100]
}


# Step 7: Perform Grid Search
xgb_grid = GridSearchCV(
    estimator=pipeline,
    param_grid=param_grid,
    cv=cross_validation,
    scoring=f3_scorer,
    n_jobs=-1,
    verbose=True
)

xgb_grid.fit(X_train_vec, y_train)

# Step 8: Evaluate the best model
best_xgb = xgb_grid.best_estimator_
y_pred = best_xgb.predict(X_test_vec)

print(" Best Parameters:", xgb_grid.best_params_)
print("\n Classification Report:\n", classification_report(y_test, y_pred))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

```python
import pandas as pd
import numpy as np
import re
import string
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

import nltk
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')

# Step 1: Sample DataFrame
df_new_data = pd.DataFrame({
    "Review": [
        "This product is perfect!",
        "I don't recommend this product, it doesn't work",
        "If you are looking for a computer, buy it!",
        "I bought this laptop for my brother.",
        "Breakdown after 3 weeks. Don't buy, really lousy customer service, no refund or gift code!!!",
        "Pc that supports games like League Of Legends with the RTX 3080 ti GE and the ryzen 9 3900X.",
        "The computer doesn't work, this is not acceptable at all!! I want a PAYBACK!"
    ]
})

# Step 2: Custom Text Preprocessing Function
def clean_text(text):
    # Lowercase
    text = text.lower()

    # Remove URLs, HTML, mentions, hashtags, emojis
    text = re.sub(r"http\S+|www\S+|<.*?>|@\w+|#\w+|\W+", " ", text)

    # Remove digits and extra whitespace
    text = re.sub(r"\d+", " ", text)
    text = re.sub(r"\s+", " ", text).strip()

    # Tokenize
    tokens = word_tokenize(text)

    # Stopwords
    stop_words = set(stopwords.words('english'))
    # Keep these important negations/sentiment words
    stopwords_to_keep = {
        "doesn", "doesn't", "doesnt", "dont", "don't", "not", "wasn't", "wasnt",
        "aren", "aren't", "arent", "couldn", "couldn't", "couldnt", "didn", "didn't",
        "didnt", "hadn", "hadn't", "hadnt", "hasn", "hasn't", "hasnt", "haven't",
        "havent", "isn", "isn't", "isnt", "mightn", "mightn't", "mightnt", "mustn",
        "mustn't", "mustnt", "needn", "needn't", "neednt", "shan", "shan't", "shant",
        "shouldn", "shouldn't", "shouldnt", "weren", "weren't", "werent", "won",
        "won't", "wont", "wouldn", "wouldn't", "wouldnt", "good", "bad", "worst",
        "wonderfull", "best", "better"
    }
    stop_words = stop_words - stopwords_to_keep

    tokens = [t for t in tokens if t not in stop_words]

    # Lemmatize
    lemmatizer = WordNetLemmatizer()
    tokens = [lemmatizer.lemmatize(t) for t in tokens]

    return " ".join(tokens)

# Apply cleaning
df_new_data["Review"] = df_new_data["Review"].apply(clean_text)

# Step 3: Vectorization (use previously fitted TF-IDF vectorizer)
list_new_data = df_new_data["Review"].tolist()
new_data_vec = vectorizer.transform(list_new_data)
```
